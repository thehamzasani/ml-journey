```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
print("\n" + "="*50)
print("MINI TRAINING LOOP — the blueprint of all ML")
print("="*50)

np.random.seed(42)
X = np.random.uniform(0, 5, 20)
y = 3 * X + 2 + np.random.normal(0, 0.5, 20)
w = 0.0
b = 0.0
lr = 0.01
n = len(X)
print("n = ", n)
loss = []
for epoch in range(300):
    y_pred = w*X+b
    error = y_pred - y
    # print("error", error)
    loss = np.mean(error**2)
    # print("loss = " ,loss)
    losses.append(loss)
    dw = (2/n) * np.dot(error, X)
    db = (2/n) * np.sum(error)
    w = w - lr * dw
    b = b - lr * db
    if epoch % 60 == 0:
        print(f"Epoch {epoch:3d} | loss={loss:.4f} | "
              f"w={w:.3f} | b={b:.3f}")

print(f"\nFinal:  w={w:.3f} (true=3.0)  b={b:.3f} (true=2.0)")
print(f"Model discovered the true values just by taking steps!")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Data + learned line
axes[0].scatter(X, y, color='steelblue', alpha=0.7, label='Data')
x_line = np.linspace(0, 5, 100)
axes[0].plot(x_line, w*x_line+b, 'red', lw=2,
             label=f'Learned: y={w:.2f}x+{b:.2f}')
axes[0].plot(x_line, 3*x_line+2, 'green', lw=1.5,
             linestyle='--', label='True: y=3x+2')
axes[0].legend(); axes[0].set_title('Learned line vs true line')

# Loss curve
axes[1].plot(losses, color='tomato', lw=2)
axes[1].set_title('Loss dropping over 300 epochs')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Loss')
axes[1].grid(alpha=0.3)

plt.tight_layout(); plt.show()
```
