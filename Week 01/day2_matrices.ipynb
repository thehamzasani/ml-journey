```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
```

```python
print("A =\n", A)
print("\nB =\n", B)
print("\nA @ B =\n", A @ B)         
print("\nA transpose =\n", A.T)
```

```python
v = np.array([1, 0])
print("\nA applied to [1,0] =", A @ v)
print("A applied to [0,1] =", A @ np.array([0, 1]))
```

```python
def show_transform(matrix, title, color):
    fig, ax = plt.subplots(figsize=(5, 5))
    
    # Draw transformed grid lines
    for x in np.linspace(-2, 2, 9):
        p1 = matrix @ np.array([x, -2])
        p2 = matrix @ np.array([x,  2])
        ax.plot([p1[0], p2[0]], [p1[1], p2[1]],
                color=color, alpha=0.5, lw=0.8)
    
    for y in np.linspace(-2, 2, 9):
        p1 = matrix @ np.array([-2, y])
        p2 = matrix @ np.array([ 2, y])
        ax.plot([p1[0], p2[0]], [p1[1], p2[1]],
                color=color, alpha=0.5, lw=0.8)
    
    # Draw basis vectors
    origin = np.array([0, 0])
    i_hat = matrix @ np.array([1, 0])
    j_hat = matrix @ np.array([0, 1])
    
    ax.annotate('', xy=i_hat, xytext=origin,
                arrowprops=dict(arrowstyle='->', color='red', lw=2.5))
    ax.annotate('', xy=j_hat, xytext=origin,
                arrowprops=dict(arrowstyle='->', color='blue', lw=2.5))
    
    ax.set_xlim(-3, 3); ax.set_ylim(-3, 3)
    ax.axhline(0, color='black', lw=0.5)
    ax.axvline(0, color='black', lw=0.5)
    ax.set_aspect('equal')
    ax.set_title(title)
    ax.grid(False)
    plt.show()
```

```python
stretch = np.array([[2, 0], [0, 1]])
angle = np.radians(45)
rotation = np.array([[np.cos(angle), -np.sin(angle)],
                     [np.sin(angle),  np.cos(angle)]])
shear = np.array([[1, 1], [0, 1]])
```

```python
show_transform(stretch,  'Stretch — pulls space wider',   '#1D9E75')
show_transform(rotation, 'Rotation — spins space 45°',    '#7F77DD')
show_transform(shear,    'Shear — slides space sideways', '#D85A30')
```

```python
my_matrix = np.array([[1, 0],
                      [0, -1]])

show_transform(my_matrix, 'My experiment', 'steelblue')
print("Where does [1,0] go?", my_matrix @ np.array([1, 0]))
print("Where does [0,1] go?", my_matrix @ np.array([0, 1]))
```
