```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
v1 = np.array([2, 3])
v2 = np.array([1, -1])
```

```python
print("v1 =", v1)
print("v2 =", v2)
print("v1 + v2 =", v1 + v2)     
print("v1 - v2 =", v1 - v2)       
print("v1 × 3 =", v1 * 3)          
print("dot product =", np.dot(v1, v2)) 
print("length of v1 =", np.linalg.norm(v1))  
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

ax.annotate('', xy=v1, xytext=[0,0],
            arrowprops=dict(arrowstyle='->', color='red', lw=2))
ax.text(v1[0]+0.1, v1[1]+0.1, 'v1 = [2,3]', color='red', fontsize=12)

ax.annotate('', xy=v2, xytext=[0,0],
            arrowprops=dict(arrowstyle='->', color='blue', lw=2))
ax.text(v2[0]+0.1, v2[1]+0.1, 'v2 = [1,-1]', color='blue', fontsize=12)

v_sum = v1 + v2
ax.annotate('', xy=v_sum, xytext=[0,0],
            arrowprops=dict(arrowstyle='->', color='green', lw=2))
ax.text(v_sum[0]+0.1, v_sum[1]+0.1, 'v1+v2', color='green', fontsize=12)

ax.set_xlim(-1, 4); ax.set_ylim(-2, 4)
ax.axhline(0, color='black', lw=0.5)
ax.axvline(0, color='black', lw=0.5)
ax.grid(alpha=0.3)
ax.set_aspect('equal')
ax.set_title('Vectors as arrows in space')
plt.show()
```

```python
a = np.array([1, 0])   
b = np.array([1, 0])   
c = np.array([0, 1])   
d = np.array([-1, 0]) 
```

```python
print("\nSame direction:    dot =", np.dot(a, b))  
print("Perpendicular:     dot =", np.dot(a, c))   
print("Opposite:          dot =", np.dot(a, d))   
```
