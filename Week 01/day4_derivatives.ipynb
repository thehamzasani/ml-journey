```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_range = np.linspace(-3, 3, 100)
y_range = x_range ** 2
dy_range = 2 * x_range  # derivative

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(x_range, y_range, 'steelblue', lw=2, label='f(x) = x²')
axes[0].set_title('The function')
axes[0].set_xlabel('x'); axes[0].set_ylabel('f(x)')
axes[0].grid(alpha=0.3); axes[0].legend()

axes[1].plot(x_range, dy_range, 'tomato', lw=2, label="f'(x) = 2x")
axes[1].axhline(0, color='black', lw=0.5)
axes[1].set_title('The derivative (slope at each point)')
axes[1].set_xlabel('x'); axes[1].set_ylabel("f'(x)")
axes[1].grid(alpha=0.3); axes[1].legend()

plt.tight_layout(); plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x_range, y_range, 'steelblue', lw=2.5, label='f(x) = x²')
for x0, color in [(2, 'red'), (0, 'green'), (-2, 'orange')]:
    slope = 2 * x0
    tangent = slope * (x_range - x0) + x0**2
    ax.plot(x_range, tangent, color=color, lw=1.5, linestyle='--',
            label=f'x={x0}: slope={slope}')
    ax.scatter([x0], [x0**2], color=color, s=80, zorder=5)

ax.set_ylim(-1, 9)
ax.set_title('Tangent lines at different points')
ax.legend(); ax.grid(alpha=0.3)
plt.show()
```
