```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
flips = np.random.choice(['H', 'T'], size=10000)
heads = np.sum(flips == 'H')
print(f"Heads: {heads}/10000 = {heads/10000:.3f}")
print(f"Expected: 0.500")
```

```python
n = 2000
flips = np.random.randint(0, 2, size=n)
running_avg = np.cumsum(flips) / np.arange(1, n+1)
```

```python
plt.figure(figsize=(10, 4))
plt.plot(running_avg, color='steelblue', lw=1.5)
plt.axhline(0.5, color='red', linestyle='--',
            label='True probability = 0.5')
plt.xlabel('Number of flips')
plt.ylabel('Proportion of heads')
plt.title('Wild at first — then stabilizes near 0.5')
plt.legend()
plt.show()
```

```python
trials = 10000
first  = np.random.randint(0, 2, size=trials)
second = np.random.randint(0, 2, size=trials)

first_was_heads  = second[first == 1]
print(f"\nP(second=H | first=H) = {np.mean(first_was_heads):.3f}")
print("Should be ~0.5 — coin has no memory")
```
